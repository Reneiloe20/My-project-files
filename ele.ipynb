```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.array([0, 0.1, 0.176, 0.2, 0.3])
y = np.array([0.140, 0.268, 0.456, 0.736, 0.940])

# Line through 1st and 4th points
m = (0.736 - 0.140) / (0.2 - 0)
c = 0.140  # intercept since it passes through (0, 0.140)

# Generate line
x_fit = np.linspace(0, 0.32, 100)
y_fit = m * x_fit + c

# Plot
plt.scatter(x, y, label="Data points")
plt.plot(x_fit, y_fit)

plt.xlabel("V_OUT (Multimeter)")
plt.ylabel("V_OUT (Oscilloscope)")

plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
I = np.array([1.1e-6,2.9e-6,5.1e-6,7.2e-6,9.1e-6])
V = np.array([1,1.64,2.48,3.44,4.40])
slope=(I*V).sum() /(I**2).sum()
I_fit=np.linspace(min(I),max(I),100)
V_fit=slope*I_fit
plt.figure()
plt.plot(I,V,marker='o')
plt.plot(I_fit,V_fit,label='Straight line set',color='red')
plt.xlabel('Current ')
plt.ylabel('Voltage(V)')
plt.title('I versus V')
plt.grid(True)
plt.legend()
plt.show()
```
